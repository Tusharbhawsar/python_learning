```python
import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from PIL import Image
import os

class CustomImageDataset(Dataset):
    def __init__(self, image_dir, transform=None):
        self.image_dir = image_dir
        self.transform = transform
        self.image_files = [f for f in os.listdir(image_dir) if f.endswith(".jpg")]

    def __len__(self):
        return len(self.image_files)

    def __getitem__(self, idx):
        img_path = os.path.join(self.image_dir, self.image_files[idx])
        image = Image.open(img_path).convert("RGB")
        label = 0  # Replace with actual label logic

        if self.transform:
            image = self.transform(image)
        
        return image, label

# Transforms
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5]*3, std=[0.5]*3)
])

# Dataset and Loader
dataset = CustomImageDataset(image_dir='path/to/images', transform=transform)
loader = DataLoader(dataset, batch_size=32, shuffle=True, num_workers=4)

# Iterate through batches
for images, labels in loader:
    print(images.shape)  # torch.Size([32, 3, 224, 224])
```

```python
#Python class methods ka first argument hamesha self hota hai, jo object ko refer karta hai.
class Pali:
    @staticmethod                    #without defining the self parameter
    def check_palidrome(stng):     
        a = stng
        b = a[::-1]

        is_pali = True
        for i in range(len(a)):
            if a[i] != b[i]:
                is_pali = False
                break

        if is_pali:
            print("palidrome")
        else:
            print("no palidrome")

th = Pali()
th.check_palidrome("madam")
```

```python
class Calculater:
    def __init__(self,a:int,b:int):
        self.a = a
        self.b = b

    def add (self):
        return self.a + self.b
    
    def mul(self):
        return self.a * self.b
    
    def sub(self):
        return self.a - self.b
    
    def divide(self):
        if self.b !=0:
            return self.a / self.b
        else:
            return "can'nt divisible by zero"
```
